## 01. What is OOP?

**Object-Oriented Programming (OOP)** is a programming paradigm that organises software design around **data** (objects) rather than functions and logic. An object is a data field that has unique attributes and behaviour.

The main principles of OOP:
1. **Encapsulation** - the internal representation of an object is hidden from the outside. Only necessary information is exposed; the rest stays private, protecting data integrity.
2. **Inheritance** - a new class (subclass) inherits properties and behaviour from an existing class (parent). It can override or extend the parent's functionality, promoting code reusability.
3. **Polymorphism** - objects of different classes can be treated as objects of a common superclass, letting a single interface represent different underlying forms.
4. **Abstraction** - hide complex implementation details and show only the necessary features, so developers work with high-level concepts.

OOP promotes code organisation, modularity, and maintainability. Popular OOP languages include Java, C++, Python, and Ruby.


## 02. Classes and Objects

- A **class** is a **blueprint** for creating objects. It defines the attributes and methods that objects created from it will have.
- An **object** is an **instance** of a class - a specific realisation with its own unique state and behaviour.

Think of a class as the blueprint for a house, and an object as a specific house built from that blueprint. The class defines the structure (number of rooms, roof type), while each object has its own characteristics (colour, size).

Example - the `House` class:

```
|       Class: House          |
| Attributes:                 |
| - number_of_rooms           |
| - type_of_roof              |
| - color                     |
| - size                      |
| Methods:                    |
| - build()                   |
| - paint()                   |
```

A class itself does not consume memory, but each object created from it allocates memory for its own attributes.


### Defining a Class and Creating Objects

```python
class ClassName:
    # class body
    # attributes and methods
```

```python
object_name = ClassName()
```

Attributes are accessed with **dot notation**: `object.attribute`. Methods are called with `object.method()`.

Here is a `Student` class with shared class attributes and methods that use `self` to reach the instance.


````python
class Student:
    # class attributes (shared by all instances)
    school_name = "ABC High School"
    name = ""
    age = 0
    grade = ""

    # method to simulate studying
    def study(self):
        print(f"{self.name} is studying.")

    # method to simulate taking an exam
    def take_exam(self):
        print(f"{self.name} is taking an exam.")

    # method to get the student's grade
    def get_grade(self):
        return self.grade


# Creating an object (instance) of the Student class
student1 = Student()
student1.name = "Alice"
student1.age = 20
student1.grade = "A"
print(f"Student Name: {student1.name}")
print(f"Student Age: {student1.age}")
print(f"Student Grade: {student1.get_grade()}")
student1.study()
student1.take_exam()

# Creating another object (instance) of the Student class
student2 = Student()
student2.name = "Bob"
student2.age = 22
student2.grade = "B"
print(f"Student Name: {student2.name}")
print(f"Student Age: {student2.age}")
print(f"Student Grade: {student2.get_grade()}")
student2.study()
student2.take_exam()

print(student1)  # <__main__.Student object at 0x...>
print(student2)  # <__main__.Student object at 0x...>
````

## 03. Attributes and Methods

- **Attributes** are variables that belong to an object.
- **Methods** are functions that belong to an object.

Both are accessed and modified with **dot notation**:

```python
class Person:
    name = "John"  # This is an attribute

    def __init__(self, name, age):
        self.name = name  # instance attribute

    def greet(self):  # This is a method
        print(f"Hello, my name is {self.name} and I am {self.age} years old.")

person1 = Person("Alice", 30)
person1.greet()  # Call the greet method using dot notation
```

Note: Python's convention is to define the constructor as `def __init__(self, ...)`.


````python
class Person:
    name = "John"  # This is an attribute

    def __init__(self, name, age):
        self.name = name  # This is an instance attribute
        self.age = age    # This is another instance attribute

    def greet(self):  # This is a method
        print(f"Hello, my name is {self.name} and I am {self.age} years old.")


person1 = Person("Alice", 30)
print(person1.name)  # Access the attribute using dot notation
print(person1.age)   # Access the age attribute using dot notation
person1.greet()      # Call the greet method using dot notation
````

## 04. Constructors

A **constructor** is a special method that is **automatically called** when an object of a class is created. In Python the constructor is named `__init__`.

- It is used to **initialize the attributes** of the object.
- It takes at least one parameter, usually named **`self`**, which refers to the instance being created.
- The `self` parameter lets the constructor access and modify the object's attributes.

```python
class Student:
    def __init__(self, name, age, grade):
        self.name = name
        self.age = age
        self.grade = grade

student1 = Student("Alice", 20, "A")
```

- Constructors can include **default values** for parameters, so objects can be created with fewer arguments.
- The first parameter of `__init__` is always `self` (by convention), and additional parameters set up the object's initial state.


````python
class Student:
    def __init__(self, name, age=18, grade="A"):
        self.name = name
        self.age = age
        self.grade = grade


student1 = Student("Alice", 20, "A")
print(student1.name)  # Output: Alice
print(student1.age)   # Output: 20
print(student1.grade) # Output: A

# Default values are used when arguments are omitted
student2 = Student("Bob")
print(student2.name)  # Output: Bob
print(student2.age)   # Output: 18
print(student2.grade) # Output: A
````

## 06. Attributes

Attributes are variables that belong to an object. There are two kinds:

- **Instance attributes** - specific to each instance, defined inside the constructor with `self`.
- **Class attributes** - shared by all instances, defined outside any method, accessed via the class name.

```python
class Student:
    school = "ABC School"  # class attribute

    def __init__(self, name, age):
        self.name = name  # instance attribute
        self.age = age    # instance attribute
```

Attributes can be added and modified after creation (`student1.grade = "A"`), and deleted with `del`.

**Lookup priority** when accessing an attribute:
1. Instance attributes
2. Class attributes
3. Parent class attributes
4. Built-in attributes

If not found anywhere, Python raises an `AttributeError`.


````python
class Student:
    school = "ABC School"  # class attribute

    def __init__(self, name, age):
        self.name = name  # instance attribute
        self.age = age    # instance attribute


student1 = Student("Alice", 20)
print(student1.name)  # Output: Alice
print(student1.age)   # Output: 20

student1.grade = "A"  # adding a new attribute to the object
print(student1.grade)  # Output: A

student1.age = 21  # modifying the age attribute
print(student1.age)   # Output: 21

del student1.age  # deleting the age attribute
try:
    print(student1.age)  # raises AttributeError since age was deleted
except AttributeError as e:
    print(e)  # Output: 'Student' object has no attribute 'age' 
````

## 07. Methods

**Methods** are functions defined inside a class. They are defined with `def` (indented within the class) and their first parameter is always **`self`**, which refers to the instance the method is called on.

- Methods can take **additional parameters** besides `self` to act on the instance.
- Methods can **return values** using `return`.
- Methods are called with **dot notation**: `student1.display_info()`.

Beyond regular instance methods, Python supports:
- **Class methods** - use the `@classmethod` decorator and take `cls` as the first parameter. They access and modify class attributes.
- **Static methods** - use the `@staticmethod` decorator and take no special first parameter. They act as utility functions that need no instance or class data.


````python
class Student:
    school_name = "ABC School"

    def __init__(self, name, age):  # Instance method
        self.name = name
        self.age = age

    def display_info(self):  # Instance method
        print(f"Name: {self.name}, Age: {self.age}")

    def update_age(self, new_age):  # Instance method with extra parameter
        self.age = new_age

    def get_info(self):  # Instance method returning a value
        return f"Name: {self.name}, Age: {self.age}"

    @classmethod
    def get_school_name(cls):  # Class method
        return cls.school_name

    @staticmethod
    def is_adult(age):  # Static method
        return age >= 18


student1 = Student("Alice", 20)
student1.display_info()   # Output: Name: Alice, Age: 20
student1.update_age(21)
student1.display_info()   # Output: Name: Alice, Age: 21
info = student1.get_info()
print(info)               # Output: Name: Alice, Age: 21

print(Student.get_school_name())  # Output: ABC School
print(Student.is_adult(20))       # Output: True
print(Student.is_adult(15))       # Output: False
````

## 08. Practice Problem - Product Store

Design and create an online store for **Products** (`name`, `price`):
- Track the **total number of products** being created.
- Create a **static method** to calculate the discount on each product based on a percentage parameter.

This practice ties together instance attributes, a class counter, a class method, and a static method.


````python
class Product:
    count = 0

    def __init__(self, name, price):
        self.name = name
        self.price = price
        Product.count += 1

    def get_info(self):
        return f"Product Name: {self.name}, Price: ${self.price}"

    @classmethod
    def get_count(cls):
        return f"Total products in the store: {cls.count}"

    @staticmethod
    def calculate_discount(price, discount_percent):
        discount_amount = price * (discount_percent / 100)
        discounted_price = price - discount_amount
        return discounted_price


p1 = Product("Laptop", 50_000)
p2 = Product("Mobile", 20_000)
p3 = Product("Pen", 15)

print(p1.get_info())
print(p2.get_info())
print(p3.get_info())
print(Product.get_count())

discounted_price = Product.calculate_discount(p1.price, 10)
print(f"Discounted price of {p1.name} after 10% discount: ${discounted_price}")
````

## 09. OOP Pillars

Object-Oriented Programming is built on **four pillars**:

1. **Encapsulation** - bundling data and the methods that operate on it within a single unit (a class), and restricting direct access to some components to prevent accidental modification.

2. **Abstraction** - hiding the complex implementation details and showing only the necessary features, letting you focus on interactions at a higher level.

3. **Inheritance** - one class inherits properties and behaviours (fields and methods) from another, promoting code reusability and establishing a hierarchy.

4. **Polymorphism** - different classes can be treated as instances of the same class through a common interface, so methods can be used interchangeably.


## 10. Encapsulation

**Encapsulation** bundles data and the methods that operate on it inside a single unit and **restricts direct access** to some components.

Python achieves this with **access modifiers**:
- **Public** - accessible from anywhere. All members are public by default (`balance`).
- **Protected** - intended for use inside the class and its subclasses. Indicated by a single underscore (`_balance`).
- **Private** - intended for use inside the class only. Indicated by a double underscore (`__account_number`). Private members are **name-mangled** to prevent accidental outside access.

Encapsulation gives us:
1. **Data hiding** - internal state is hidden behind a controlled interface.
2. **Modularity** - internal implementation can change without affecting users of the class.
3. **Maintenance** - hidden internals reduce unintended interference.

Private attributes can still be reached via name mangling (`object._ClassName__attribute`), but this is not recommended.


````python
class BankAccount:
    def __init__(self, owner, balance=0, account_number=None):
        self.owner = owner             # Public attribute
        self._balance = balance        # Protected attribute
        self.__account_number = account_number  # Private attribute

    def deposit(self, amount):
        if amount > 0:
            self._balance += amount
            print(f"Deposited {amount}. New balance: {self._balance}")
        else:
            print("Deposit amount must be positive.")

    def withdraw(self, amount):
        if 0 < amount <= self._balance:
            self._balance -= amount
            print(f"Withdrew {amount}. New balance: {self._balance}")
        else:
            print("Withdrawal amount must be positive and less than or equal to the balance.")

    def get_balance(self):
        return self._balance

    def get_account_number(self):
        return self.__account_number


# Example usage
account = BankAccount("Alice", 1000, "123456789")
account.deposit(500)   # Deposited 500. New balance: 1500
account.withdraw(200)  # Withdrew 200. New balance: 1300
print(account.get_balance())  # 1300
print(account.get_account_number())  # 123456789

# Private attributes are name-mangled, not directly accessible:
# print(account.__account_number)  # AttributeError

# But can be reached via name mangling (not recommended)
print(account._BankAccount__account_number)  # 123456789

# Protected attributes can be accessed but are intended for internal use
print(account._balance)  # 1300
````

## 11. Inheritance

**Inheritance** allows a new class (the **child/subclass**) to inherit attributes and methods from an existing class (the **parent/superclass**).

```python
class ParentClass:
    # parent class code

class ChildClass(ParentClass):
    # child class code
```

- It **reuses** attributes and methods of an existing class.
- It can **add features** to a class without modifying it.
- It creates a natural **hierarchical relationship** between classes.
- The **`super()`** function calls the parent class constructor so the child can initialise inherited attributes:

```python
class ChildClass(ParentClass):
    def __init__(self, args):
        super().__init__(args)  # Call the parent class constructor
        # Additional initialization for the child class
```


````python
class Employee:
    start_time = "9:00 AM"
    end_time = "5:00 PM"

    def change_start_time(self, new_start_time):
        self.start_time = new_start_time

    def change_end_time(self, new_end_time):
        self.end_time = new_end_time


class Teacher(Employee):
    def __init__(self, subject):
        self.subject = subject


class Manager(Employee):
    def __init__(self, department):
        self.department = department


t1 = Teacher("Math")
m1 = Manager("HR")
print("Teacher", t1.subject, t1.start_time, t1.end_time)
print("Manager", m1.department, m1.start_time, m1.end_time)


# Inheritance with super() and method overriding
class Animal:
    def __init__(self, name, species):
        self.name = name
        self.species = species

    def make_sound(self):
        pass


class Dog(Animal):
    def __init__(self, name, breed):
        super().__init__(name, "Canine")
        self.breed = breed

    def make_sound(self):
        return "Woof!"


class Cat(Animal):
    def __init__(self, name, color):
        super().__init__(name, "Feline")
        self.color = color

    def make_sound(self):
        return "Meow!"


dog = Dog("Buddy", "Golden Retriever")
cat = Cat("Whiskers", "Orange")
print(dog.name, dog.species, dog.breed, dog.make_sound())
print(cat.name, cat.species, cat.color, cat.make_sound())
````

## 12. Types of Inheritance

Python supports five types of inheritance:

1. **Single Inheritance** - a child class inherits from one parent class.
2. **Multiple Inheritance** - a child class inherits from more than one parent class (can cause ambiguity with conflicting members).
3. **Multilevel Inheritance** - a child inherits from a parent, which in turn inherits from another parent, forming a chain.
4. **Hierarchical Inheritance** - multiple child classes inherit from a single parent class, forming a tree-like structure.
5. **Hybrid Inheritance** - a combination of two or more of the above types.


### Single and Multiple Inheritance

**Single inheritance** reuses the members of one parent. **Multiple inheritance** lets a child inherit from several parents at once, combining their features.


````python
# Single Inheritance
class Employee:
    start_time = "9:00 AM"
    end_time = "5:00 PM"

class AdminStaff(Employee):
    def __init__(self, role):
        self.role = role

admin = AdminStaff("HR Manager")
print(admin.role, admin.start_time, admin.end_time)


# Multiple Inheritance
class Teacher:
    def __init__(self, subject, salary):
        self.subject = subject
        self.salary = salary

class Student:
    def __init__(self, cgpa):
        self.cgpa = cgpa

class TeachingAssistant(Teacher, Student):
    def __init__(self, subject, salary, name, cgpa):
        Teacher.__init__(self, subject, salary)  # Initialize Teacher attributes
        Student.__init__(self, cgpa)             # Initialize Student attributes
        self.name = name

ta = TeachingAssistant("Math", 30000, "Alice", 3.8)
print(ta.name, ta.subject, ta.salary, ta.cgpa)
````

### Multilevel, Hierarchical, and Hybrid Inheritance

- **Multilevel** - a chain of classes, e.g. `Employee` → `AdminStaff` → `Accountant`.
- **Hierarchical** - several children of one parent, e.g. `Developer` and `Designer` both extend `Employee`.
- **Hybrid** - a combination, such as multilevel plus single inheritance.


````python
# Multilevel Inheritance (Employee -> AdminStaff -> Accountant)
class Employee:
    start_time = "9:00 AM"

class AdminStaff(Employee):
    def __init__(self, role):
        self.role = role

class Accountant(AdminStaff):
    def __init__(self, salary, role):
        super().__init__(role)  # Call the parent constructor
        self.salary = salary

accountant = Accountant(50000, "Accountant")
print(accountant.role, accountant.salary, accountant.start_time)


# Hierarchical Inheritance
class Developer(Employee):
    def __init__(self, programming_language):
        self.programming_language = programming_language

class Designer(Employee):
    def __init__(self, design_tool):
        self.design_tool = design_tool

developer = Developer("Python")
designer = Designer("Adobe Photoshop")
print(developer.programming_language, developer.start_time)
print(designer.design_tool, designer.start_time)


# Hybrid Inheritance (multilevel + single)
class Manager(Employee):
    def __init__(self, department):
        self.department = department

class TeamLead(Manager):
    def __init__(self, department, team_size):
        super().__init__(department)
        self.team_size = team_size

team_lead = TeamLead("IT", 5)
print(team_lead.department, team_lead.team_size, team_lead.start_time)
````

## 13. Abstraction

**Abstraction** hides implementation details and shows only the functionality. It lets us focus on **what** an object does, not **how** it does it.

In Python, abstraction is achieved with **abstract classes**:
- An abstract class **cannot be instantiated**; it is meant to be subclassed.
- It can declare **abstract methods** (declared but not implemented) that subclasses must implement.

Use the `abc` module:

```python
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass
```

- `ABC` stands for Abstract Base Class.
- The `@abstractmethod` decorator marks a method as abstract.


````python
from abc import ABC, abstractmethod

# Define an abstract class
class Shape(ABC):
    @abstractmethod
    def area(self):
        pass


# Subclass must implement the abstract method
class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius ** 2


circle = Circle(5)
print(circle.area())  # Output: 78.5
````

## 14. Polymorphism

**Polymorphism** means "many forms": *poly* (many) + *morph* (form). It allows objects of different classes to be treated as objects of a common superclass - a single interface representing different underlying types.

In Python it appears as:
1. **Method Overriding** - a subclass provides its own implementation of a method already defined in its superclass, keeping the same interface but different behaviour.
2. **Duck Typing** - the type of an object is determined by its behaviour rather than its class: *"If it looks like a duck and quacks like a duck, it must be a duck."*
3. **Operator Overloading** - operators behave differently by operand type, e.g. `+` adds numbers but concatenates strings. Custom classes can overload operators with special methods like `__add__` and `__str__`.


````python
# Method Overriding (inheritance)
class Employee:
    def get_designation(self):
        print("designation = Employee")

class Teacher(Employee):
    def get_designation(self):
        print("designation = Teacher")

t1 = Teacher()
t1.get_designation()


# Duck Typing
class Dog:
    def speak(self):
        print("Bark")

class Cat:
    def speak(self):
        print("Meow")

class Robot:
    def speak(self):
        print("Beep Boop")

def make_it_speak(entity):
    entity.speak()  # does not care about type

for e in [Dog(), Cat(), Robot()]:
    make_it_speak(e)


# Operator Overloading
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Point(self.x + other.x, self.y + other.y)

    def __str__(self):
        return f"({self.x}, {self.y})"

p1 = Point(1, 2)
p2 = Point(3, 4)
p3 = p1 + p2  # Uses the overloaded + operator
print(p3)     # Output: (4, 6)
````

**Q1.** Create a `BankAccount` class with attributes `account_number`, `owner_name`, and `balance`. Add methods to deposit, withdraw, and check balance.

**Q2.** Create a `Book` class with attributes `title`, `author`, and a list of `reviews`. Add methods to add a review, count reviews, and display all reviews.

**Q3.** Create a `Student` class with private attributes `_name`, `_roll_no`, and `_marks`. Provide getter and setter methods with validation (marks cannot be negative, roll number must be between 1 and 100, name cannot be empty).


````python
# Q1 Solution: BankAccount
class BankAccount:
    def __init__(self, account_number, owner_name, balance):
        self.account_number = account_number
        self.owner_name = owner_name
        self.balance = balance

    def deposit(self, amount):
        if amount >= 0:
            self.balance += amount
            print("Deposit amount:", amount, "New Balance:", self.balance)
        else:
            print("Deposit amount cannot be negative")

    def withdraw(self, amount):
        if amount >= 0:
            if self.balance >= amount:
                self.balance -= amount
                print("Withdraw amount:", amount, "New Balance:", self.balance)
            else:
                print("You don't have enough balance.")
        else:
            print("Withdraw amount cannot be negative")

    def check_balance(self):
        print("Current balance:", self.balance)
        return self.balance

a1 = BankAccount("7323254823", "Zeanur Rahaman Zeon", 30200)
a1.check_balance()
a1.deposit(200)
a1.withdraw(200)
a1.check_balance()


# Q2 Solution: Book
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author
        self.reviews = []

    def add_review(self, review):
        self.reviews.append(review)
        print("Review added:", review)

    def count_reviews(self):
        print("Total reviews:", len(self.reviews))
        return len(self.reviews)

    def display_reviews(self):
        print("Reviews for", self.title, "by", self.author)
        for review in self.reviews:
            print("-", review)

b1 = Book("The Great Gatsby", "F. Scott Fitzgerald")
b1.add_review("A classic novel.")
b1.add_review("Great character development.")
b1.count_reviews()
b1.display_reviews()


# Q3 Solution: Student with validation
class Student:
    def __init__(self, name, roll_no, marks):
        self.__name = None
        self.__roll_no = None
        self.__marks = None
        self.set_name(name)
        self.set_roll_no(roll_no)
        self.set_marks(marks)

    def set_name(self, name):
        if name.strip() != "":
            self.__name = name
            print("Name set to:", name)
        else:
            print("Name cannot be empty.")

    def set_roll_no(self, roll_no):
        if 1 <= roll_no <= 100:
            self.__roll_no = roll_no
            print("Roll number set to:", roll_no)
        else:
            print("Roll number must be between 1 and 100.")

    def set_marks(self, marks):
        if marks >= 0:
            self.__marks = marks
            print("Marks set to:", marks)
        else:
            print("Marks cannot be negative.")

    def get_name(self):
        return self.__name

    def get_roll_no(self):
        return self.__roll_no

    def get_marks(self):
        return self.__marks

s1 = Student("Alice", 10, 85)
s1.set_marks(90)
print("Name:", s1.get_name())
print("Roll Number:", s1.get_roll_no())
print("Marks:", s1.get_marks())
````
