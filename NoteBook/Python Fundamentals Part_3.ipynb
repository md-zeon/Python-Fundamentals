# Python Fundamentals: Part 3

This part explores the **data structures** that Python uses to store and organise data: strings, lists, tuples, dictionaries, and sets.

## What you will learn
- Strings: creation, immutability, slicing, and formatting
- The rich set of built-in string methods
- Lists: indexing, mutation, slicing, sorting, and looping
- Tuples: immutability, unpacking, and conversions
- Dictionaries: key-value pairs and their methods
- Sets: uniqueness and set operations
- Solving real problems with data structures
