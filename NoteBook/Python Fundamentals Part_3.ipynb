## 01. Strings

A **string** is a sequence of characters enclosed in quotes. It can be defined using single quotes, double quotes, or triple quotes (for multi-line text).

- Strings are **immutable**: once created, their characters cannot be changed in place.
- They are one of the most common data types and can contain letters, numbers, symbols, and whitespace.
- Strings can be manipulated with many built-in methods and operators, which makes them powerful for handling text.
- **Escape characters** like `\n` (newline) and `\'` (quote) let you embed special characters.
- **Raw strings** (`r"..."`) keep backslashes literal, which is handy for file paths like `C:\Users\...`.


Strings can be created in several ways: with single, double, or triple quotes, escape characters, raw strings, concatenation, and the various formatting techniques (`f-strings`, `format()`, and the `%` operator).


```python
# Using single quotes
string1 = 'Hello, World!'
print(string1)

# Using double quotes
string2 = "Python is great!"
print(string2)

# Using triple quotes (multi-line strings)
string3 = '''This is a multi-line string.'''
print(string3)

# Using escape characters
string5 = 'It\'s a nice day!'
print(string5)

# Using raw strings (backslashes stay literal)
string6 = r'C:\Users\Username\Documents'
print(string6)

# Using string concatenation
string7 = string1 + " " + string2
print(string7)

# Using f-strings
name = "Alice"
age = 30
string8 = f"My name is {name} and I am {age} years old."
print(string8)

# Using the format() method
string9 = "My name is {} and I am {} years old.".format(name, age)
print(string9)

# Using the % operator
string10 = "My name is %s and I am %d years old." % (name, age)
print(string10)

# Using string literals
string11 = "This is a string literal."
print(string11)
```

### Common String Methods

Strings come with methods for **changing case**, **searching**, and **splitting**:
- `upper()` / `lower()` - convert case
- `title()` / `capitalize()` - title case and first-letter capitalisation
- `replace(old, new)` - replace substrings
- `split(sep)` - split into a list
- `strip()` - remove leading/trailing whitespace
- `find(sub)` - index of first occurrence
- `count(sub)` - number of occurrences
- `startswith(prefix)` / `endswith(suffix)` - prefix/suffix checks
- `center(width)` / `ljust(width)` / `rjust(width)` / `zfill(width)` - padding and alignment
- `encode()` - encode the string into bytes


```python
string1 = 'Hello, World!'

# Case conversion
print("Using upper():", string1.upper())
print("Using lower():", string1.lower())
print("Using title():", string1.title())
print("Using capitalize():", string1.capitalize())

# Search and replace
print("Using replace():", string1.replace("World", "Python"))
print("Using find():", string1.find("World"))
print("Using count():", string1.count("o"))

# Splitting and stripping
print("Using split():", string1.split(","))
print("Using strip():", string1.strip())

# Prefix and suffix checks
print("Using startswith():", string1.startswith("Hello"))
print("Using endswith():", string1.endswith("!"))
```
