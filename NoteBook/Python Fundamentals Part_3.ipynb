## 01. Strings

A **string** is a sequence of characters enclosed in quotes. It can be defined using single quotes, double quotes, or triple quotes (for multi-line text).

- Strings are **immutable**: once created, their characters cannot be changed in place.
- They are one of the most common data types and can contain letters, numbers, symbols, and whitespace.
- Strings can be manipulated with many built-in methods and operators, which makes them powerful for handling text.
- **Escape characters** like `\n` (newline) and `\'` (quote) let you embed special characters.
- **Raw strings** (`r"..."`) keep backslashes literal, which is handy for file paths like `C:\Users\...`.


Strings can be created in several ways: with single, double, or triple quotes, escape characters, raw strings, concatenation, and the various formatting techniques (`f-strings`, `format()`, and the `%` operator).


````python
# Using single quotes
string1 = 'Hello, World!'
print(string1)

# Using double quotes
string2 = "Python is great!"
print(string2)

# Using triple quotes (multi-line strings)
string3 = '''This is a multi-line string.'''
print(string3)

# Using escape characters
string5 = 'It\'s a nice day!'
print(string5)

# Using raw strings (backslashes stay literal)
string6 = r'C:\Users\Username\Documents'
print(string6)

# Using string concatenation
string7 = string1 + " " + string2
print(string7)

# Using f-strings
name = "Alice"
age = 30
string8 = f"My name is {name} and I am {age} years old."
print(string8)

# Using the format() method
string9 = "My name is {} and I am {} years old.".format(name, age)
print(string9)

# Using the % operator
string10 = "My name is %s and I am %d years old." % (name, age)
print(string10)

# Using string literals
string11 = "This is a string literal."
print(string11)
````

### Common String Methods

Strings come with methods for **changing case**, **searching**, and **splitting**:
- `upper()` / `lower()` - convert case
- `title()` / `capitalize()` - title case and first-letter capitalisation
- `replace(old, new)` - replace substrings
- `split(sep)` - split into a list
- `strip()` - remove leading/trailing whitespace
- `find(sub)` - index of first occurrence
- `count(sub)` - number of occurrences
- `startswith(prefix)` / `endswith(suffix)` - prefix/suffix checks
- `center(width)` / `ljust(width)` / `rjust(width)` / `zfill(width)` - padding and alignment
- `encode()` - encode the string into bytes


````python
string1 = 'Hello, World!'

# Case conversion
print("Using upper():", string1.upper())
print("Using lower():", string1.lower())
print("Using title():", string1.title())
print("Using capitalize():", string1.capitalize())

# Search and replace
print("Using replace():", string1.replace("World", "Python"))
print("Using find():", string1.find("World"))
print("Using count():", string1.count("o"))

# Splitting and stripping
print("Using split():", string1.split(","))
print("Using strip():", string1.strip())

# Prefix and suffix checks
print("Using startswith():", string1.startswith("Hello"))
print("Using endswith():", string1.endswith("!"))
````

### Padding, Alignment, and Encoding

The alignment and padding methods shape how a string is displayed within a field of a given **width**:
- `center(width)` - centres the string in a field
- `ljust(width)` - left-justifies the string
- `rjust(width)` - right-justifies the string
- `zfill(width)` - pads the left with zeros
- `encode()` - converts the string into **bytes**


````python
string1 = 'Hello, World!'

# Padding and alignment
print("Using center():", string1.center(20))
print("Using ljust():", string1.ljust(20))
print("Using rjust():", string1.rjust(20))
print("Using zfill():", string1.zfill(20))

# Encoding
print("Using encode():", string1.encode())
````

### String Validation Methods

These methods **test** the characters in a string and return `True` or `False`:
- `isalpha()` - only alphabetic characters
- `isdigit()` / `isnumeric()` / `isdecimal()` - numeric characters
- `isalnum()` - alphanumeric characters
- `isspace()` - only whitespace
- `islower()` / `isupper()` - all lower/upper case
- `isidentifier()` - a valid Python identifier
- `isprintable()` - all characters printable
- `isascii()` - all characters ASCII


````python
string1 = 'Hello, World!'

print("Using isalpha():", string1.isalpha())
print("Using isdigit():", string1.isdigit())
print("Using isalnum():", string1.isalnum())
print("Using isspace():", string1.isspace())
print("Using islower():", string1.islower())
print("Using isupper():", string1.isupper())
print("Using isnumeric():", string1.isnumeric())
print("Using isdecimal():", string1.isdecimal())
print("Using isidentifier():", string1.isidentifier())
print("Using isprintable():", string1.isprintable())
print("Using isascii():", string1.isascii())
````

## 02. String Slicing

**Slicing** extracts a portion of a string by specifying a range of indices:

```python
string[start:stop:step]
```

- **start** - where the slice begins (inclusive). Defaults to `0`.
- **stop** - where the slice ends (exclusive). Defaults to the length of the string.
- **step** - the interval between characters. Defaults to `1`. A negative step slices in reverse.


````python
string1 = 'Hello, World!'

# Substring from index 0 to 5 (exclusive)
substring1 = string1[0:5]  # 'Hello'
print(string1, "sliced from index 0 to 5:", substring1)

# Substring from index 7 to the end
substring2 = string1[7:]  # 'World!'
print(string1, "sliced from index 7 to the end:", substring2)

# Substring from the beginning to index 5 (exclusive)
substring3 = string1[:5]  # 'Hello'
print(string1, "sliced from the beginning to index 5:", substring3)

# Every second character
substring4 = string1[::2]  # 'Hlo ol!'
print(string1, "sliced with a step of 2:", substring4)

# Reverse the string
substring5 = string1[::-1]  # '!dlroW ,olleH'
print(string1, "sliced in reverse order:", substring5)

# From index 2 to 10 with a step of 3
substring6 = string1[2:10:3]  # 'loW'
print(string1, "sliced from index 2 to 10 with a step of 3:", substring6)

# From index 5 to 0 in reverse
substring7 = string1[5:0:-1]  # ', ol'
print(string1, "sliced from index 5 to 0 in reverse order:", substring7)

# From index 10 to 2 in reverse
substring8 = string1[10:2:-1]  # 'roW ,o'
print(string1, "sliced from index 10 to 2 in reverse order:", substring8)
````

## 03. String Formatting

String formatting embeds expressions inside string literals. Python offers three main techniques:

1. **`format()` method** - uses `{}` placeholders. Supports positional arguments, named placeholders, and formatting options like width, precision, and alignment.
2. **f-strings** (Python 3.6+) - prefix the string with `f` and embed expressions directly in `{}`. The most concise and readable approach.
3. **`%` operator** - the older style, using format specifiers like `%s` (string) and `%d` (integer). Still supported for backward compatibility.


````python
name = "Alice"
age = 30

# format() method with positional placeholders
print("My name is {} and I am {} years old.".format(name, age))

# format() method with named placeholders
print("My name is {name} and I am {age} years old.".format(name=name, age=age))

# format() method with indexed placeholders
print("My name is {0} and I am {1} years old. {0} is a common name.".format(name, age))

# f-strings
name = "Bob"
age = 25
print(f"My name is {name} and I am {age} years old.")

# % operator (old-style formatting)
name = "Charlie"
age = 35
print("My name is %s and I am %d years old." % (name, age))
````

## 04. Lists

A **list** is a built-in data structure that stores a collection of items. Lists are:
- **Ordered** - items keep their insertion order.
- **Mutable** - items can be changed, added, or removed after creation.
- **Heterogeneous** - can hold different types, including other lists.

Lists are defined with square brackets `[]`.

```python
my_list = [1, "Hello", 3.14, [2, 4, 6]]
```

- Access items by **index**, starting at `0`; negative indices count from the end (`-1` is the last item).
- Because lists are mutable, you can assign to an index: `my_list[0] = 42`.


````python
my_list = [1, "Hello", 3.14, [2, 4, 6]]

# Accessing elements by index
print(my_list[0])  # 1
print(my_list[1])  # Hello
print(my_list[2])  # 3.14
print(my_list[3])  # [2, 4, 6]

# Negative indexing (from the end)
print(my_list[-1])  # [2, 4, 6]
print(my_list[-2])  # 3.14

# Lists are mutable
my_list[0] = 42
print(my_list[0])  # 42
````

### Adding and Removing Elements

Common list operations:
- **Add**: `append(item)` adds to the end; `insert(index, item)` inserts at a position.
- **Remove**: `remove(item)` removes by value; `pop(index)` removes by index (and returns it).
- **Length**: `len()` gives the number of elements.


````python
my_list = [42, "Hello", 3.14, [2, 4, 6]]

# Append an item to the end
my_list.append("New Item")
print(my_list)  # [42, 'Hello', 3.14, [2, 4, 6], 'New Item']

# Remove an item by value
my_list.remove("Hello")
print(my_list)  # [42, 3.14, [2, 4, 6], 'New Item']

# Remove an item by index
my_list.pop(1)
print(my_list)  # [42, [2, 4, 6], 'New Item']

# Get the length of the list
print(len(my_list))  # 3
````

### Slicing, Concatenation, and Nesting

- **Slicing** `list[start:end]` creates a new list with a subset of the elements; negative indices slice from the end.
- **Concatenation** `list1 + list2` joins two lists; **repetition** `list * n` repeats the list.
- The **`in` operator** checks membership.
- Lists can be **nested** (lists inside lists), accessed with multiple indices like `nested[1][0]`.
- Lists may contain **duplicate** elements.


````python
my_list = [42, [2, 4, 6], "New Item"]

# Slicing
sliced_list = my_list[0:2]
print(sliced_list)  # [42, [2, 4, 6]]

# Negative slicing from the end
sliced_list = my_list[-2:]
print(sliced_list)  # [[2, 4, 6], 'New Item']

# Concatenation with +
another_list = [7, 8, 9]
combined_list = my_list + another_list
print(combined_list)  # [42, [2, 4, 6], 'New Item', 7, 8, 9]

# Repetition with *
repeated_list = my_list * 2
print(repeated_list)

# Membership with in
print(42 in my_list)      # True
print("Hello" in my_list) # False

# Nested lists
nested_list = [1, [2, 3], [4, 5, 6]]
print(nested_list)      # [1, [2, 3], [4, 5, 6]]
print(nested_list[1][0])  # 2
print(nested_list[2][1])  # 5

# Lists can contain duplicates
duplicate_list = [1, 2, 2, 3, 4, 4, 4]
print(duplicate_list)  # [1, 2, 2, 3, 4, 4, 4]
````

### Sorting, Reversing, and Looping

- **`sort()`** sorts in place (ascending by default; pass `reverse=True` for descending). All elements must be of the same type to compare.
- **`reverse()`** reverses the order in place; **`clear()`** empties the list.
- **`sorted()`** returns a new sorted list, leaving the original untouched.
- Loop over a list with a **`for` loop**, a **`while` loop**, or `for` with **`range()`**.
- **List comprehensions** build a new list with an expression: `[x**2 for x in my_list]`.


````python
numbers_list = [3, 1, 4, 1, 5, 9]

# sort() in ascending order
numbers_list.sort()
print(numbers_list)  # [1, 1, 3, 4, 5, 9]

# sort() in descending order
numbers_list.sort(reverse=True)
print(numbers_list)  # [9, 5, 4, 3, 1, 1]

# sorted() returns a new list without modifying the original
original_list = [3, 1, 4, 1, 5, 9]
sorted_list = sorted(original_list)
print(sorted_list)   # [1, 1, 3, 4, 5, 9]
print(original_list) # [3, 1, 4, 1, 5, 9]

my_list = [1, 2, 3, 4, 5]

# Loop with a for loop
print("Using for loop:")
for element in my_list:
    print(element, end=" ")  # 1 2 3 4 5

# Loop with a while loop
print("\nUsing while loop:")
index = 0
while index < len(my_list):
    element = my_list[index]
    print(element, end=" ")  # 1 2 3 4 5
    index += 1

# Loop with for and range()
print("\nUsing for loop with range:")
for index in range(len(my_list)):
    element = my_list[index]
    print(element, end=" ")  # 1 2 3 4 5

# List comprehension
squared_list = [x**2 for x in my_list]
print("\nSquared list using list comprehension:")
print(squared_list)  # [1, 4, 9, 16, 25]
````

## 05. Tuples

A **tuple** is an ordered collection of items that is **immutable** - once created, its contents cannot be changed. Tuples are defined with parentheses `()` and can hold elements of different types, including other tuples.

```python
my_tuple = (1, "Hello", 3.14, [1, 2, 3], (4, 5))
```

Special tuple forms:
- **Single-element tuple**: `(5,)` - the comma is essential! Without it, `(5)` is just the integer `5`.
- **Nested tuples**: `(1, 2, (3, 4), 5)`
- **Empty tuple**: `()`
- Tuples can hold mixed types and duplicates.

Common operations: `len()`, `in` membership, `+` concatenation, `*` repetition, **unpacking** (`a, b, c = tuple`), slicing, `count()`, `index()`, `sorted()` (returns a list), and conversion with `list()` / `tuple()`.


````python
my_tuple = (1, "Hello", 3.14, [1, 2, 3], (4, 5))

# Accessing tuple elements
print(my_tuple[0])  # 1
print(my_tuple[1])  # Hello
print(my_tuple[2])  # 3.14

# Length and membership
print(len(my_tuple))   # 5
print(3.14 in my_tuple)  # True

# Concatenation and repetition
tuple1 = (1, 2, 3)
tuple2 = (4, 5, 6)
concatenated_tuple = tuple1 + tuple2
print(concatenated_tuple)  # (1, 2, 3, 4, 5, 6)
repeated_tuple = tuple1 * 3
print(repeated_tuple)  # (1, 2, 3, 1, 2, 3, 1, 2, 3)

# Unpacking
a, b, c = tuple1
print(a, b, c)  # 1 2 3

# Slicing
sliced_tuple = concatenated_tuple[2:5]
print(sliced_tuple)  # (3, 4, 5)

# count() and index()
duplicate_tuple = (1, 2, 2, 3, 4, 4, 5)
print(duplicate_tuple.count(2))  # 2
print(duplicate_tuple.index(4))  # 4

# sorted() returns a new list
unsorted_tuple = (3, 1, 4, 2)
print(sorted(unsorted_tuple))  # [1, 2, 3, 4]

# Conversions
tuple_to_list = list(my_tuple)
print(tuple_to_list)
list_to_tuple = tuple(tuple_to_list)
print(list_to_tuple)

# Single-element tuple needs a comma
single_tuple = (5,)
print(type(single_tuple))  # <class 'tuple'>
not_single_tuple = (5)
print(type(not_single_tuple))  # <class 'int'>

# Empty tuple
empty_tuple = ()
print(type(empty_tuple))  # <class 'tuple'>
````

### Looping Through Tuples

Tuples work with `for` loops just like lists. The **`enumerate()`** function gives both the index and the value on each iteration, which is very handy when the position matters.

```python
for index, element in enumerate(my_tuple):
    print(f"Index: {index}, Element: {element}")
```

You can also iterate with a `while` loop and convert a tuple into a new list with a comprehension.


````python
my_tuple = (1, "Hello", 3.14, [1, 2, 3], (4, 5))

# For loop over a tuple
for element in my_tuple:
    print(element)

# enumerate() gives index and value
for index, element in enumerate(my_tuple):
    print(f"Index: {index}, Element: {element}")

# While loop over a tuple
index = 0
while index < len(my_tuple):
    print(my_tuple[index])
    index += 1

# List comprehension from a tuple
squared_tuple = (1, 2, 3, 4, 5)
squared_list = [x**2 for x in squared_tuple]
print(squared_list)  # [1, 4, 9, 16, 25]
````

## 06. Dictionaries

A **dictionary** is a collection of **key-value pairs**. Each key is unique and maps to a specific value.

```python
person = {
    "name": "Zeanur Rahaman Zeon",
    "age": 23,
    "hobbies": ["coding", "traveling", "cooking"],
}
```

- Dictionaries are **mutable** - you can add, change, or remove pairs.
- Keys must be **immutable** (strings, numbers, tuples); values can be anything - lists, dictionaries, even functions.
- Since Python 3.7, dictionaries keep the **insertion order**.


````python
person = {
    "name": "Zeanur Rahaman Zeon",
    "age": 23,
    "city": "Dhaka",
    "hobbies": ["coding", "traveling", "cooking"],
    "education": {
        "degree": "Bachelor's",
        "major": "Computer Science",
        "university": "University of Dhaka"
    },
    123: "This is a numeric key",
    (1, 2): "This is a tuple key",
}

# Accessing values with square brackets
print(person["name"])  # Zeanur Rahaman Zeon
print(person["age"])   # 23
print(person[123])     # This is a numeric key
print(person[(1, 2)])  # This is a tuple key

# Accessing with get() - safe, with a default
print(person.get("city"))                  # Dhaka
print(person.get("country", "Not Found"))  # Not Found

# Accessing with items()
for key, value in person.items():
    print(f"Key: {key}, Value: {value}")
````

### Dictionary Methods

Dictionaries ship with a full set of methods:
- **`keys()`**, **`values()`**, **`items()`** - views over the content
- **`update()`** - merge in another dictionary
- **`pop(key)`** - remove a key and return its value
- **`popitem()`** - remove and return an arbitrary pair
- **`clear()`**, **`copy()`**, **`get()`**
- **`setdefault()`** - return the value, inserting a default if the key is missing
- **`fromkeys()`** - build a new dictionary from an iterable of keys


````python
person = {
    "name": "Zeanur Rahaman Zeon",
    "age": 23,
    "city": "Dhaka",
    "hobbies": ["coding", "traveling", "cooking"],
    "education": {
        "degree": "Bachelor's",
        "major": "Computer Science",
        "university": "University of Dhaka"
    },
    123: "This is a numeric key",
    (1, 2): "This is a tuple key",
}

# keys(), values(), items()
print(person.keys())
print(person.values())
print(person.items())

# update() merges another dictionary
new_info = {"country": "Bangladesh", "age": 24}
person.update(new_info)
print(person)

# pop() removes a key and returns its value
age = person.pop("age")
print(age)
print(person)

# popitem() removes an arbitrary pair
key, value = person.popitem()
print(f"Removed key: {key}, value: {value}")
print(person)

# copy()
person_copy = person.copy()
print(person_copy)

# setdefault() inserts a default if the key is missing
default_value = person.setdefault("country", "Not Found")
print(default_value)  # Not Found (key was already removed by popitem)
print(person)

# fromkeys() builds a new dictionary
keys = ["name", "age", "city"]
new_dict = dict.fromkeys(keys, "Unknown")
print(new_dict)  # {'name': 'Unknown', 'age': 'Unknown', 'city': 'Unknown'}

# clear()
person.clear()
print(person)  # {}
````

## 07. Sets

A **set** is an unordered collection of **unique** elements.

```python
fruits = {"apple", "banana", "cherry", "apple"}  # duplicates removed
numbers = set([1, 2, 3, 4, 5])
empty_set = set()  # {} would make an empty dict!
```

- Sets are **mutable** (add/remove elements), but the elements themselves must be immutable.
- Sets are **unordered** - elements cannot be accessed by index.
- Key methods: `add()`, `remove()` (raises if missing), `discard()` (no error), `pop()`, `clear()`, `copy()`.
- **Set operations**: `union()`, `intersection()`, `difference()`, `symmetric_difference()`.
- **Relationships**: `issubset()`, `issuperset()`, `isdisjoint()`.
- A **`frozenset`** is an immutable set.


````python
# Creating sets
fruits = {"apple", "banana", "cherry", "apple"}
print(fruits)  # {'banana', 'cherry', 'apple'}

numbers = set([1, 2, 3, 4, 5, 2, 3])
print(numbers)  # {1, 2, 3, 4, 5}

empty_set = set()
print(empty_set)  # set()

# add, remove, discard, pop, clear
fruits = {"apple", "banana", "cherry"}
fruits.add("orange")
print(fruits)
fruits.remove("banana")
print(fruits)
fruits.discard("grape")  # no error, "grape" is not present
print(fruits)
popped_fruit = fruits.pop()  # arbitrary element
print(popped_fruit)
fruits.clear()
print(fruits)  # set()

# Set operations
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}
print(set_a.union(set_b))                # {1, 2, 3, 4, 5, 6}
print(set_a.intersection(set_b))         # {3, 4}
print(set_a.difference(set_b))           # {1, 2}
print(set_a.symmetric_difference(set_b)) # {1, 2, 5, 6}

# Subset, superset, disjoint
set_c = {1, 2}
print(set_c.issubset(set_a))   # True
print(set_a.issuperset(set_c)) # True
set_d = {7, 8}
print(set_a.isdisjoint(set_d)) # True

# Copy and frozenset
set_e = set_a.copy()
print(set_e)
frozen_set = frozenset([1, 2, 3, 4])
print(frozen_set)
````

## 08. Practice - Student Enrollments

A classic data-structure problem: given a list of `(student, subject)` tuples, we want to:
1. List all **unique courses** (a job for a set).
2. List students **enrolled in English**.
3. Build a dictionary mapping each **student to a set of courses**.

This combines tuples (the pairs), lists (the input), a dictionary (student → courses), and sets (unique courses).


````python
info = [
    ("Alice", "Math"),
    ("Bob", "English"),
    ("Alice", "Science"),
    ("Charlie", "Math"),
    ("Bob", "Math"),
    ("Charlie", "English"),
]

# List all unique courses
unique_courses = set()
print("Unique courses:", end=" ")
for student, course in info:
    unique_courses.add(course)
print(unique_courses)

# List students enrolled in English
print("Students enrolled in English: ", end=" ")
for student, course in info:
    if course == "English":
        print("[" + student + "]", end=" ")
print()

# Create a dictionary (student, set of courses)
enrollments = {}
for student, course in info:
    if student not in enrollments:
        enrollments[student] = set()
    enrollments[student].add(course)

print("Enrollments:", enrollments)
````

## 09. Assignment Problems

**Q1.** Ask the user for a string and check whether it is a **palindrome** (same forwards and backwards - e.g. "madam", "racecar").
> Hint: a palindrome string equals its reversed version. You can reverse it manually with a loop, or use slicing `[::-1]`, or compare with two pointers.

**Q2.** Given a list of integers, compute the **average** of all numbers in the list.

**Q3.** Input two lists of integers from the user. **Merge** them into one list and **sort** the result.
> Example: `list1 = [1, 2, 7]`, `list2 = [3, 4, 5]` → `result = [1, 2, 3, 4, 5, 7]`


````python
# Q1 Solution: palindrome check (three approaches)
def answer_q1():
    # Method 1: reverse manually with a loop
    string = input("Enter a string: ")
    reversed_string = ""
    for c in string:
        reversed_string = c + reversed_string
    if string == reversed_string:
        print("Given String is a palindrome")
    else:
        print("Given String is not a palindrome")

    # Method 2: slicing
    string = input("Enter a string: ")
    reversed_string = string[::-1]
    if string == reversed_string:
        print("Given String is a palindrome")
    else:
        print("Given String is not a palindrome")

    # Method 3: two pointers
    string = input("Enter a string: ")
    i, j = 0, len(string) - 1
    is_palindrome = True
    while i < j:
        if string[i] != string[j]:
            is_palindrome = False
            break
        i += 1
        j -= 1
    if is_palindrome:
        print("Given String is a palindrome")
    else:
        print("Given String is not a palindrome")

# Q2 Solution: average of a list
def answer_q2():
    numbers = [1, 2, 3, 4, 5]
    average = sum(numbers) / len(numbers)
    print("Average:", average)

# Q3 Solution: merge and sort two lists
def answer_q3():
    list1 = list(map(int, input("Enter the first list of integers (space-separated): ").split()))
    list2 = list(map(int, input("Enter the second list of integers (space-separated): ").split()))
    merged_list = list1 + list2
    merged_list.sort()
    print("Merged and sorted list:", merged_list)
````
