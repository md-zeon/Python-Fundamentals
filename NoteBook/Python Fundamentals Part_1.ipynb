## 01. First Program

A *program* is a sequence of instructions that a computer executes. The simplest instruction in Python is the built-in function `print()`, which sends output to the screen.

Key ideas:

- `print()` is a **built-in function** — it is always available and needs no imports.
- The text inside the parentheses is a **string**, which is a sequence of characters.
- `print()` can take several arguments separated by commas; they are printed one after another separated by a space.
- `"\n"` is an **escape sequence** that inserts a new line.
- The `end` parameter decides what is printed *after* the output (the default is a newline `"\n"`).
- `print()` can also print the result of an expression.


````python
print("Hello Zeon!")
# print() outputs the message inside the parentheses to the screen

print("Hello World!", "With \nPython!")
# Multiple arguments are printed in one line, separated by a space
# The "\n" escape sequence starts a new line in the output

print("Hello Zeon!", end=" ")
print("Welcome to Python programming!")
# Setting end=" " keeps the next print on the same line

print(10 + 5)
# print() can also display the result of an expression
````

## 02. Variables and Identifiers

A **variable** is a container for storing data values. You create a variable and give it a value with the assignment operator `=`.

```python
age = 30
name = "John"
```

Rules and facts about variables:

- A variable name is called an **identifier**. It is the name given to a variable, function, class, or other object.
- An identifier must start with a letter (`a-z`, `A-Z`) or an underscore (`_`).
- It can be followed by letters, digits (`0-9`), or underscores.
- It cannot be a reserved keyword in Python (such as `if`, `for`, `while`).
- Python is **dynamically typed**: you do not declare a type. The type is decided at runtime from the value you assign, so a variable can change type.
- Choose **meaningful names** that describe the data they hold.

### How variables are stored in memory

When you create a variable, the name is a *reference* to a location in memory where the value is stored. Assigning a new value updates the reference to point to a new location, which is why a variable can hold different types and change throughout a program.

```python
x = 10       # x references the memory holding 10
x = "Hello"  # x now references a new location holding "Hello"
```


````python
age = 30
name = "John"

# Multiple variables can be assigned in a single line
x, y, z = 1, 2, 3

# Variables can be reassigned to new values at any time
age = 31   # now holds 31
name = "Jane"  # now holds "Jane"

# Valid identifiers
my_variable = 10
_name = "Alice"
age2 = 25

# Invalid identifiers (uncomment to see the errors):
# 2age = 30        # cannot start with a digit
# my-variable = 5  # cannot contain a hyphen
````

## 03. Data Types

Python comes with several built-in **data types**:

| Type      | Description                                    | Example                       |
|-----------|------------------------------------------------|-------------------------------|
| `int`     | Whole numbers                                  | `1`, `42`, `-5`               |
| `float`   | Numbers with a decimal point                   | `3.14`, `-0.001`, `2.0`       |
| `str`     | Sequence of characters in quotes               | `"Hello"`, `'World'`, `"123"` |
| `bool`    | Logical `True` or `False`                      | `True`, `False`               |
| `list`    | Ordered, mutable collection                    | `[1, 2, 3]`, `["a", "b"]`     |
| `tuple`   | Ordered, immutable collection                  | `(1, 2, 3)`                   |
| `dict`    | Collection of key-value pairs                  | `{"name": "Alice", "age": 30}`|
| `set`     | Unordered collection of unique items           | `{1, 2, 3}`                   |
| `NoneType`| Represents the absence of a value              | `None`                        |

The built-in function `type()` returns the type of any value or variable.


````python
name = "Zeon"
age = 22
PI = 3.14

print(type(name))  # <class 'str'>
print(type(age))   # <class 'int'>
print(type(PI))    # <class 'float'>

# Using different data types
my_height = 5.7            # float
is_student = True          # bool
hobbies = ["reading", "swimming", "coding"]  # list
address = ("123 Main St", "City", "Country") # tuple
person_info = {"name": "Zeon", "age": 22}    # dict
unique_numbers = {1, 2, 3, 4, 5}             # set
my_value = None             # NoneType

print(type(my_height))
print(type(is_student))
print(type(hobbies))
print(type(address))
print(type(person_info))
print(type(unique_numbers))
print(type(my_value))
````

## 05. Naming Style Guide

There are four common naming styles for variables and functions:

1. **Camel Case** — first letter of each word is capitalised, no spaces or underscores.  
   `myVariableName`, `userProfile`, `employeeData`
2. **Pascal Case** — like Camel Case, but the *first* word is also capitalised.  
   `MyVariableName`, `UserProfile`, `EmployeeData`
3. **Snake Case** — all lowercase, words separated by underscores.  
   `my_variable_name`, `user_profile`, `employee_data`
4. **Kebab Case** — all lowercase, words separated by hyphens.  
   `my-variable-name`, `user-profile`, `employee-data`

In Python, the most widely used convention is **Snake Case** for variable names and function names.


````python
# Python convention: snake_case for variables and functions
my_variable_name = "Hello, World!"
user_profile = {"name": "Zeon", "role": "Student"}

def calculate_average(a, b):
    return (a + b) / 2

print(my_variable_name)
print(calculate_average(10, 20))
````

## 06. Operators

Operators perform operations on values. Python provides several categories of operators:

1. Arithmetic Operators: `+`, `-`, `*`, `/`, `%`, `**`, `//`
2. Comparison Operators: `==`, `!=`, `>`, `<`, `>=`, `<=`
3. Logical Operators: `and`, `or`, `not`
4. Assignment Operators: `=`, `+=`, `-=`, `*=`, `/=`, `%=`
5. Bitwise Operators: `&`, `|`, `^`, `~`, `<<`, `>>`
6. Identity Operators: `is`, `is not`
7. Membership Operators: `in`, `not in`

### Arithmetic operators

`+` adds, `-` subtracts, `*` multiplies, `/` divides (always returns a float), `%` gives the remainder, `**` raises to a power, and `//` performs **floor division** (rounds the quotient down to the nearest whole number).

### Comparison operators

Comparison operators compare two values and return a boolean (`True` or `False`).


````python
a = 10
b = 5

# Arithmetic operators
print("Addition:", a + b)          # 15
print("Subtraction:", a - b)       # 5
print("Multiplication:", a * b)    # 50
print("Division:", a / b)          # 2.0  (always a float)
print("Modulus:", a % b)           # 0
print("Exponentiation:", a ** b)   # 100000
print("Floor Division:", a // b)   # 2

# Comparison operators
x = 10
y = 20
print("Equal to:", x == y)              # False
print("Not equal to:", x != y)          # True
print("Greater than:", x > y)           # False
print("Less than:", x < y)              # True
print("Greater than or equal:", x >= y) # False
print("Less than or equal:", x <= y)    # True
````

### Logical operators

`and`, `or`, and `not` combine boolean values to build more complex conditions.

### Assignment operators

Assignment operators update a variable with a value. `x += 3` is a shortcut for `x = x + 3`.


````python
# Logical operators
p = True
q = False
print("Logical AND:", p and q)   # False
print("Logical OR:", p or q)     # True
print("Logical NOT:", not p)     # False

# Assignment operators
x = 5
x += 3
print("After += 3:", x)   # 8
x -= 2
print("After -= 2:", x)   # 6
x *= 4
print("After *= 4:", x)   # 24
x /= 6
print("After /= 6:", x)   # 4.0
x %= 3
print("After %= 3:", x)   # 1.0
````

### Bitwise operators

Bitwise operators work on the **binary representation** of integers.

- `&` bitwise AND, `|` bitwise OR, `^` bitwise XOR, `~` bitwise NOT
- `<<` shifts bits left (multiplies by powers of two), `>>` shifts bits right

### Identity operators

`is` and `is not` check whether two variables **refer to the same object in memory** (not whether their values are equal).

### Membership operators

`in` and `not in` check whether a value appears inside a sequence such as a list or string.


````python
# Bitwise operators
a = 5   # binary 0101
b = 3   # binary 0011
print("Bitwise AND:", a & b)   # 1  (0001)
print("Bitwise OR:", a | b)    # 7  (0111)
print("Bitwise XOR:", a ^ b)   # 6  (0110)
print("Bitwise NOT:", ~a)      # -6 (inverts all bits)
print("Left Shift:", a << 1)   # 10 (1010)
print("Right Shift:", a >> 1)  # 2  (0010)

# Identity operators
x = 5
y = 5
z = 10
print("x is y:", x is y)          # True
print("x is z:", x is z)          # False
print("x is not z:", x is not z)  # True

# Membership operators
my_list = [1, 2, 3, 4, 5]
print("1 in my_list:", 1 in my_list)         # True
print("6 in my_list:", 6 in my_list)         # False
print("6 not in my_list:", 6 not in my_list) # True
````

## 07. Operator Precedence

**Operator precedence** is the priority given to operators inside an expression. Operators with higher precedence are evaluated first. When operators have the same precedence, they are evaluated **left to right**.

From highest to lowest precedence:

1. Parentheses `()`
2. Exponentiation `**`
3. Multiplication `*`, Division `/`, Floor Division `//`, Modulus `%`
4. Addition `+`, Subtraction `-`
5. Comparison operators `==`, `!=`, `>`, `<`, `>=`, `<=`
6. Logical `not`
7. Logical `and`
8. Logical `or`

Use parentheses to change the order of evaluation and make expressions clearer.
