## 01. Modules and Packages

A **module** is a Python file that contains definitions and statements - functions, classes, and variables. Modules let you organise code into reusable pieces and split a large program across several files.

Importing a module:
- `import module` - imports the whole module; members are accessed as `module.member`
- `from module import name` - imports only specific names
- `import module as alias` - imports the module under a shorter name

Python ships with a rich **standard library** of modules, including `math`, `os`, `random`, `datetime`, `json`, `csv`, and `re`.

A **package** is a folder that contains one or more modules together with an `__init__.py` file. Packages group related modules, e.g. `import package.module`.

**Note:** the code examples below reference a module named `helpers.py` that you can create in the same folder as the notebook.


### Creating Modules and the name-main Guard

Every `.py` file is a potential module. `__name__` is a special variable: it is `"__main__"` when a file runs directly, and the module's name when it is imported. The `if __name__ == "__main__":` guard runs code only when the file is executed directly.


```python
import math as m
print(m.sqrt(9))  # Output: 3.0

from math import pi
print(pi)  # Output: 3.141592653589793

# Every Python file can be a module. Save this as "helpers.py":
#   def greet(name):
#       return f"Hello, {name}!"
#
# then import it with:
#   from helpers import greet
#   print(greet("Alice"))

# In a notebook __name__ is "__main__"
print(__name__)  # Output: __main__

def main():
    print("Running as the main program.")

if __name__ == "__main__":
    main()
```

## 02. Variable Scope

The **scope** of a variable is the part of the program where it can be accessed. Python resolves names using the **LEGB** rule:
1. **L**ocal - inside the current function
2. **E**nclosing - inside enclosing (outer) functions
3. **G**lobal - at the top level of the module
4. **B**uilt-in - Python's built-in names

- A variable assigned inside a function is **local** by default.
- The `global` keyword lets a function read and modify a **global** variable.
- The `nonlocal` keyword lets a nested function modify a variable in the **enclosing** function.


```python
total = 100  # global variable

def add_to_total(amount):
    global total  # modify the global variable
    total += amount

add_to_total(50)
print(total)  # Output: 150


def outer():
    message = "Hello"  # enclosing variable

    def inner():
        nonlocal message  # modify the enclosing variable
        message = "Hi"

    inner()
    print(message)  # Output: Hi

outer()
```

## 03. Regular Expressions

A **regular expression** (regex) is a pattern that describes a set of strings. Python's `re` module works with regex patterns to search, validate, and transform text.

Common functions:
- `re.search(pattern, text)` - finds the first match anywhere in the text
- `re.match(pattern, text)` - matches from the start of the text
- `re.findall(pattern, text)` - returns a list of all matches
- `re.sub(pattern, replacement, text)` - replaces every match
- `re.split(pattern, text)` - splits the text at every match

Common pattern elements:
- `\d` - a digit
- `\w` - a word character (letter, digit, or underscore)
- `\s` - a whitespace character
- `.` - any character (except newline)
- `*` - zero or more, `+` - one or more, `?` - zero or one
- `{n,m}` - between n and m repetitions
- `[...]` - a set of characters, `^` and `$` - anchors for start and end


```python
import re

phone = "Call me at 017-1234-5678 or 018-8765-4321"

# Find all digits
print(re.findall(r"\d+", phone))
# Output: ['017', '1234', '5678', '018', '8765', '4321']

# Search for a phone number pattern
match = re.search(r"\d{3}-\d{4}-\d{4}", phone)
print(match.group())  # Output: 017-1234-5678

# Validate an email address
email_pattern = r"^[\w\.-]+@[\w\.-]+\.\w+$"
print(bool(re.match(email_pattern, "user@example.com")))  # True
print(bool(re.match(email_pattern, "not-an-email")))      # False

# Replace phone numbers
print(re.sub(r"\d{3}-\d{4}-\d{4}", "[PHONE]", phone))
```

## 04. datetime Module

The `datetime` module provides classes for working with dates and times:
- `datetime.now()` - the current date and time
- `date.today()` - today's date
- `datetime(year, month, day, hour, minute)` - a specific date and time
- `timedelta` - a duration that can be added to or subtracted from dates
- `strftime()` - format a date/time object into a string
- `strptime()` - parse a string into a date/time object

Common format codes for `strftime`: `%Y` (year), `%m` (month), `%d` (day), `%H` (hour), `%M` (minute), `%S` (second), `%A` (weekday), `%B` (month name).


```python
from datetime import datetime, date, timedelta

now = datetime.now()
print(now)          # Output: current date and time

today = date.today()
print(today)        # Output: current date

# Formatting a datetime as a string
print(now.strftime("%Y-%m-%d %H:%M:%S"))
print(now.strftime("%A, %d %B %Y"))

# Parsing a string into a datetime
parsed = datetime.strptime("2026-08-11", "%Y-%m-%d")
print(parsed)  # Output: 2026-08-11 00:00:00

# Working with time deltas
print(today + timedelta(days=30))   # 30 days from today
print(today + timedelta(weeks=2))   # 2 weeks from today

# Creating a specific date and time
appointment = datetime(2026, 12, 25, 10, 30)
print(appointment)  # Output: 2026-12-25 10:30:00
```

## 05. random Module

The `random` module generates pseudo-random numbers and picks random items:
- `random()` - a float between 0 and 1
- `randint(a, b)` - a random integer between a and b (inclusive)
- `choice(sequence)` - a random element from a sequence
- `sample(sequence, k)` - k unique random elements
- `shuffle(sequence)` - shuffles a list in place
- `seed()` - makes the random sequence reproducible


```python
import random

# A random float between 0 and 1
print(random.random())

# A random integer (like rolling a die)
print(random.randint(1, 6))

# Pick a random element
colors = ["red", "green", "blue"]
print(random.choice(colors))

# Pick 6 unique lottery numbers
print(random.sample(range(1, 50), 6))

# Shuffle a list in place
cards = list(range(1, 11))
random.shuffle(cards)
print(cards)

# Reproducible randomness with a seed
random.seed(42)
print(random.randint(1, 100))  # Always the same result after seed(42)
```
