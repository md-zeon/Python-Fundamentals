## 01. File I/O Overview

**File I/O (Input/Output)** means reading data from files and writing data to files using Python. It is essential for data storage, manipulation, and retrieval. Python provides built-in functions to work with files, and the built-in `open()` function returns a file object you can use to read or write data.

```python
file = open("filename", "mode")
```

Common file modes:

| Mode | Meaning | Description |
|------|---------|-------------|
| `r` | Read | Opens the file for reading (default). Error if the file doesn't exist. |
| `w` | Write | Creates a new file or overwrites an existing file. |
| `a` | Append | Adds content at the end of the file. |
| `r+` | Read + Write | File must exist. |
| `w+` | Write + Read | Overwrites an existing file. |
| `a+` | Append + Read | Reads and adds to the end of the file. |
| `b` | Binary | For images, videos, etc. (used with other modes, e.g. `rb`). |
| `t` | Text | Text mode (default). |

Some common file operations:
- `read()` - reads the entire file
- `readline()` - reads a single line from the file
- `readlines()` - reads all lines from the file and returns a list
- `write()` - writes data to the file
- `writelines()` - writes a list of strings to the file

**Note:** if the file does not exist and you open it in `r` mode, a `FileNotFoundError` is raised. Opening a file in `w` mode truncates it (all existing content is deleted). Always close a file after you finish with it to free up system resources.


## 02. Reading from a File

Once a file is opened for reading, its contents can be read in several ways:
- `read()` - reads the entire file as a string
- `readline()` - reads a single line from the file
- `readlines()` - reads all lines from the file and returns a list of strings

The example below reads `example.txt` with each of these methods. The `seek(0)` call moves the file pointer back to the beginning so the file can be read again.


````python
# Example of opening a file for reading
file_read = open('example.txt', 'r')

data = file_read.read()  # Reads the entire file
print(type(data))  # Output: <class 'str'>

file_read.seek(0)  # Move the file pointer back to the beginning of the file
print(data)  # Output: contents of the file as a string

line = file_read.readline()  # Reads the next line from the file
print(line)

lines = file_read.readlines()  # Reads all lines and returns a list
print(lines)

file_read.close()  # Closes the file opened for reading
````

## 03. Writing to and Closing a File

To write to a file, use the `write()` method. Opening a file in `w` mode creates a new file or overwrites an existing one; opening it in `a` mode appends to the end instead of overwriting.

After you finish working with a file, close it with the `close()` method. This releases the associated system resources and ensures that all changes are saved properly.


````python
# Example of opening a file for writing
file_write = open('example_write.txt', 'w')

# Writing to a file
file_write.write("Hello, World!\n")  # Writes a string to the file
file_write.write("This is a new line.\n")  # Writes another line to the file

# Closing the file (releases resources and saves changes)
file_write.close()

# Appending to a file
file_append = open('example_write.txt', 'a')
file_append.write("This line is appended.\n")
file_append.close()
````

## 04. The with Keyword

The `with` keyword wraps the execution of a block of code within methods defined by a **context manager**. It simplifies exception handling by managing common preparation and cleanup automatically, ensuring that resources are properly released even if an error occurs.

```python
with expression as variable:
    # code block
```

A common use case for the `with` statement is file handling:


````python
with open('example.txt', 'r') as file:
    data = file.read()
    print(data)
# The file is automatically closed after the block is exited, even if an error occurs.
````

## 05. Deleting Files

Python deletes files with the `remove()` method from the `os` module. It takes the file path as an argument and removes the specified file from the filesystem.

It is good practice to wrap `os.remove()` in a `try-except` block so the program responds gracefully when the file does not exist or another error occurs.

**Note:** once a file is deleted with `os.remove()`, it cannot be recovered through Python. Always make sure the path is correct and that you really intend to delete the file.


````python
import os

# Create a small file to delete
with open("example_delete.txt", "w") as f:
    f.write("This file will be deleted.\n")

file_path = os.getcwd() + "/example_delete.txt"
try:
    os.remove(file_path)
    print(f"{file_path} has been deleted successfully.")
except FileNotFoundError:
    print(f"{file_path} does not exist.")
except Exception as e:
    print(f"An error occurred while trying to delete {file_path}: {e}")
````

## 06. Exception Handling

An **exception** is an error that occurs while a program is running. If not handled, the program crashes. **Exception Handling** allows you to manage errors gracefully so your program continues to run.

Common exceptions:
- `ZeroDivisionError` - dividing by zero
- `NameError` - using an undefined variable
- `FileNotFoundError` - opening a missing file
- `TypeError` - using a wrong data type
- `ValueError` - a value with the right type but an invalid value

Basic syntax:

```python
try:
    # Code that may cause an error
except SomeError:
    # Code that runs if the error occurs
```

You can also **raise** specific exceptions yourself with the `raise` keyword. The `else` block runs when no error occurs, and the `finally` block always runs.


````python
# Handling a division by zero
try:
    print(10 / 0)
except ZeroDivisionError:
    print("You can't divide by zero!")

# try, except, and else blocks
try:
    num = int("42")  # valid conversion
except ValueError:
    print("Invalid input!")
else:
    print("You entered:", num)

# try, except, and finally blocks
try:
    f = open("data.txt")
    print(f.read())
except FileNotFoundError:
    print("File not found!")
finally:
    print("Execution completed.")

# Raising a specific exception
def check_age(age):
    if age < 0:
        raise ValueError("Age cannot be negative")
    return age

try:
    check_age(-5)
except ValueError as e:
    print(e)  # Output: Age cannot be negative
````
