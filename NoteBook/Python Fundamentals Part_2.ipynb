# Python Fundamentals: Part 2

This part introduces **program control flow** - how Python decides what to run, and how many times it runs.

## What you will learn
- Conditional statements: `if`, `elif`, `else`, and nested conditions
- `match`-`case` pattern matching (Python 3.10+)
- `while` and `for` loops with `range()`
- Loop control with `break` and `continue`
- Defining and calling functions
- Built-in, user-defined, lambda, recursive, and generator functions
- Applying loops and functions to real problems
