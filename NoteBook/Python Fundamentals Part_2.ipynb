## 01. Conditional Statements

Programs often need to make **decisions**. Conditional statements let the program run different blocks of code based on whether a condition is `True` or `False`.

| Statement | Purpose |
|---|---|
| `if` | Runs a block only when the condition is `True` |
| `if` / `else` | Chooses between two blocks |
| `if` / `elif` / `else` | Checks several conditions in order |
| Nested `if` | An `if` block placed inside another `if` block |

- Conditions use **comparison operators** (`==`, `!=`, `>`, `<`, `>=`, `<=`) and **logical operators** (`and`, `or`, `not`).
- An **f-string** (`f"..."`) lets you embed values directly inside a string using `{}`.
- Use the **modulo operator** (`%`) to check if a number is even or odd: `n % 2 == 0`.


````python
# if statement
x = 10
if x > 5:
    print("x is greater than 5")

# if-else statement
y = 3
if y > 5:
    print("y is greater than 5")
else:
    print("y is not greater than 5")

# if-elif-else statement
z = 7
if z < 5:
    print("z is less than 5")
elif z == 5:
    print("z is equal to 5")
else:
    print("z is greater than 5")

# Nested if statements
a = 15
if a > 10:
    if a < 20:
        print("a is between 10 and 20")
    else:
        print("a is greater than or equal to 20")

# Traffic light example
traffic_light = "red"
if traffic_light == "red":
    print("Stop")
elif traffic_light == "yellow":
    print("Get ready")
elif traffic_light == "green":
    print("Go")
else:
    print("Invalid traffic light color")

# Check if a number is even or odd
number = 8
if number % 2 == 0:
    print(f"{number} is even")  # f-string for formatted string literals
else:
    print(f"{number} is odd")  # f-string for formatted string literals

# Login system example
username = "admin"
password = "password123"
input_username = input("Enter username: ")
input_password = input("Enter password: ")
if input_username == username and input_password == password:
    print("Login successful")
else:
    print("Login failed")
````

## 02. Match-Case

The **`match`-`case`** statement (introduced in Python 3.10) provides a more elegant and readable way to handle multiple conditions compared to long `if`-`elif`-`else` chains.

- It matches a value against a set of patterns and runs the block of the first matching case.
- The underscore `_` acts as a **wildcard** that catches any value that did not match, similar to `else`.


````python
# Example of match-case statement
traffic_light = "red"
match traffic_light:
    case "red":
        print("Stop")
    case "yellow":
        print("Get ready")
    case "green":
        print("Go")
    case _:
        print("Invalid traffic light color")
````

## 03. While Loops

A **`while` loop** is a control flow statement that repeats a block of code as long as a given Boolean condition stays `True`. Once the condition becomes `False`, the loop stops.

Syntax:
```python
while condition:
    # code to be executed repeatedly
```

- A **counter variable** tracks how many times the loop has run; it must be updated inside the loop.
- **Caution:** if the condition never becomes `False`, the loop runs forever - this is called an **infinite loop**. Always make sure the loop has a way to end.


````python
# While loop that prints numbers from 1 to 5
count = 1  # counter variable for tracking the current count of the loop
while count <= 5:
    print(count, end=' ')  # Print the current count followed by a space
    count += 1  # Increment the count to avoid an infinite loop

# Infinite loop (do not run this code!)
# while True:
#     print("This will run forever!")

# To break out of an infinite loop, you can use the 'break' statement
# which we will learn about in the next section.
````

## 04. Loop Practice

A couple of quick practice problems to get comfortable with `while` loops: printing a countdown, and building a multiplication table.


````python
# Print numbers from 5 to 1
i = 5
while i >= 1:
    print(i, end=' ')
    i -= 1

# Multiplication Table
num = int(input("\nEnter a number: "))
print(f"Multiplication Table of {num}:")
i = 1
while i <= 10:
    print(f"{num} x {i} = {num * i}")
    i += 1
````

## 05. Break and Continue

The **`break`** and **`continue`** statements control the flow of loops.

- **`break`** exits the loop immediately. Execution continues with the next statement after the loop.
- **`continue`** skips the rest of the current iteration and jumps straight to the next iteration of the loop.


````python
# Example of break statement
while True:
    user_input = input("Enter a number (or 'exit' to quit): ")
    if user_input.lower() == 'exit':
        print("Exiting the loop.")
        break  # Exit the loop when the user types 'exit'
    else:
        print(f"You entered: {user_input}")

# Example of continue statement
multiples_of_num = int(input("Enter a number to find its multiples: "))
print(f"Multiples of {multiples_of_num} from 1 to 100:")
i = 1
while i <= 100:
    if i % multiples_of_num != 0:
        i += 1
        continue  # Skip the current iteration if i is not a multiple
    print(i, end=' ')
    i += 1
````

## 06. For Loops and range()

A **`for` loop** is used to iterate over a sequence (like a list, tuple, or string) or any other iterable object.

Syntax:
```python
for variable in iterable:
    # code to execute for each item in the iterable
```

- The loop variable takes the value of each item in the iterable one by one.
- The `for` loop is often combined with the `range()` function to iterate over a sequence of numbers:
```python
for variable in range(start, stop, step):
    print(variable)
```
- `start` is inclusive (default `0`), `stop` is exclusive, and `step` is the increment (default `1`).
- `for` loops can be **nested** - a `for` loop inside another `for` loop - which is useful for multi-dimensional data like lists of lists.


````python
# Example 1: Iterating over a String
string = "hello"
print("Iterating over a string:")
for var in string:
    print(var, end=' ')

# Example 2: Iterating over a List
my_list = [1, 2, 3, 4, 5]
print("\nIterating over a list:")
for num in my_list:
    print(num, end=' ')

# Example 3: Using range() in a for loop
print("\nUsing range() in a for loop:")
for i in range(5):
    print(i, end=' ')

# Example 4: Using range() with start and stop parameters
print("\nUsing range() with start and stop parameters:")
for i in range(2, 10):
    print(i, end=' ')

# Example 5: Using range() with start, stop, and step parameters
print("\nUsing range() with start, stop, and step parameters:")
for i in range(0, 20, 5):
    print(i, end=' ')
````

The `for` loop also works with **nested iteration** and the **`in` operator**.

- A nested `for` loop visits every element of a list of lists, row by row.
- The `in` operator checks whether a value is present in a string or a list. Internally it iterates over the elements and returns `True` as soon as a match is found, otherwise `False`.


````python
# Example 6: Nested for loop
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
print("Nested for loop:")
for row in matrix:
    for element in row:
        print(element, end=' ')
    print()  # Print a newline after each row

# in operator in an if statement
my_string = "hello world"
if "world" in my_string:
    print("The word 'world' is in the string.")

# the in operator also checks for elements in a list
my_list = [1, 2, 3, 4, 5]
if 3 in my_list:
    print("The number 3 is in the list.")

# Practice: count the number of i in "artificial intelligence"
word = "artificial intelligence"
count = 0
for char in word:
    if char == 'i':
        count += 1
print(f"The letter 'i' appears {count} times in the word '{word}'.")

# Practice: count the number of vowels in "artificial intelligence"
vowels = "aeiou"
count = 0
for char in word:
    if char in vowels:
        count += 1
print(f"The vowels appear {count} times in the word '{word}'.")
````

## 07. Functions

Functions are **reusable blocks of code** that perform a specific task. They help break a program into smaller, manageable pieces.

A function has two parts:
1. **Function Definition** - where the function and its parameters are declared.
2. **Function Call** - where the function's code is executed.

```python
def function_name(parameters):
    # Function body
    # Optional return statement

function_name(arguments)
```

- **`return`** hands a value back to the caller after the function finishes its task.
- **Parameters** are the placeholders declared in the definition; **arguments** are the actual values passed at call time.
- Parameters can be any data type - numbers, strings, lists, even other functions.
- Functions can define **default parameter values**, which are used when the caller omits that argument.


````python
# Example of a Function Definition and Call
def greet(name):
    print(f"Hello, {name}!")

greet("Alice")  # Output: Hello, Alice!
greet("Bob")    # Call with a different argument

# Example of a Function with a Return Value
def add(a, b):
    return a + b

result = add(5, 3)
print("The sum of 5 and 3 is:", result)  # Output: The sum of 5 and 3 is: 8

# Function with default parameter values
def average(x = 0, y = 0, z = 0):
    x = float(x)
    y = float(y)
    z = float(z)
    if (x < 0 or y < 0 or z < 0):
        return "Error: All numbers must be non-negative."
    return (x + y + z) / 3

print(average(10, 20, 30))     # Output: 20.0
print(average(-5, 10, 15))     # Output: Error: All numbers must be non-negative.
print(average(10, "20", 30))   # Output: 20.0 (the string "20" is converted to a float)
````
